##Multiple Linear Regression

```python
import pandas as pd

url = "https://raw.githubusercontent.com/CodeWithRetesh/Food_Delivery_Time_Prediction/main/dataset/Food_Delivery_Times.csv"

df = pd.read_csv(url)

print("Dataset shape:", df.shape)
df.head()
```

```python
X = df[["Distance_km","Preparation_Time_min","Courier_Experience_yrs"]].copy()
y = df["Delivery_Time_min"]
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
print("Missing values in X",X.isnull().sum())
```

```python
print("Missing values in y",y.isnull().sum())
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y , test_size=0.2,random_state=40)

median_experience = X_train["Courier_Experience_yrs"].median()
```

```python
X_train["Courier_Experience_yrs"]= X_train["Courier_Experience_yrs"].fillna(median_experience)

X_test["Courier_Experience_yrs"]= X_test["Courier_Experience_yrs"].fillna(median_experience)

print("Checking missing values after filling :",X_train.isnull().sum())
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train,y_train)
print("Model training completed!")
```

```python
y_pred = model.predict(X_test)

print("Actual values",y_test.values[:10])
print("Predicted values",y_pred[:10])
```

```python
from sklearn.metrics import mean_absolute_error,r2_score

mae = mean_absolute_error(y_test,y_pred)
r2 = r2_score(y_test,y_pred)

print("Mean absolute error :",mae)
print("R2 score :",r2)
```

Multiple Linear Regression was used to predict delivery time using distance, preparation time, and courier experience. The model achieved an MAE of approximately 8.11 minutes and an R² score of 0.73. It performed better than the Day 17 single-feature model.
